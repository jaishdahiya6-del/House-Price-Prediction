# EDA - AI House Price Prediction
Exploratory analysis on the California Housing dataset (20,640 real records).

```
import sys, os
sys.path.append(os.path.join('..', 'src'))
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
from data_loader import load_data

df = load_data()
df.head()
```

```
df.info()
df.isnull().sum()
```

```
df.describe()
```

```
plt.figure(figsize=(10,6))
sns.histplot(df['median_house_value'], kde=True)
plt.title('Target Distribution: median_house_value')
plt.show()
```

```
plt.figure(figsize=(11,9))
sns.heatmap(df.select_dtypes('number').corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```
sns.boxplot(x=df['median_income'])
plt.title('Median Income Boxplot (outlier check)')
plt.show()
```
