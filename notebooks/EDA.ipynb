# 📊 Exploratory Data Analysis (EDA)
## California Housing Dataset
This notebook explores the dataset structure, missing values, feature distributions, geographical price mapping, and correlations.

```python
import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

sys.path.append(os.path.abspath('../src'))
from data_loader import load_data

sns.set_theme(style='whitegrid')
```

### 1. Load Dataset

```python
df = load_data()
print(f'Dataset Shape: {df.shape}')
df.head()
```

### 2. Dataset Information & Missing Values

```python
df.info()
print('\nMissing Values:\n', df.isnull().sum())
df.describe()
```

### 3. Target Distribution: Median House Value

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(df['median_house_value'], kde=True, ax=axes[0], color='#2c7fb8')
axes[0].set_title('Distribution of Median House Value')
sns.boxplot(x=df['median_house_value'], ax=axes[1], color='#7fcdbb')
axes[1].set_title('Boxplot of Median House Value')
plt.tight_layout()
plt.show()
```

### 4. Categorical Feature Analysis: Ocean Proximity

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='ocean_proximity', palette='Blues_r')
plt.title('House Count by Ocean Proximity')
plt.show()
```

### 5. Geographical Distribution of Prices

```python
plt.figure(figsize=(10, 7))
scatter = plt.scatter(
    df['longitude'], df['latitude'],
    c=df['median_house_value'], cmap='jet', alpha=0.4,
    s=df['population'] / 100, label='Population'
)
plt.colorbar(scatter, label='Median House Value ($)')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Geographical Distribution of Housing Prices in California')
plt.legend()
plt.show()
```

### 6. Correlation Analysis

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, fmt='.2f', cmap='coolwarm', square=True)
plt.title('Numeric Feature Correlation Heatmap')
plt.show()
```
