# 📊 Exploratory Data Analysis (EDA)
## California Housing Dataset

This notebook provides an in-depth exploratory analysis of the **California Housing Dataset** (1990 U.S. Census). The goal of EDA is to uncover underlying structural patterns, geographical price distributions, feature correlations, and potential anomalies or outliers prior to model training.

```python
import sys
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure reusable modules from src/ are accessible
sys.path.append(os.path.abspath('../src'))
from data_loader import load_data

sns.set_theme(style='whitegrid')
%matplotlib inline
```

### 1. Dataset Ingestion
We load the cached raw dataset using `src.data_loader.load_data()`. The dataset contains 20,640 records corresponding to California census block groups.

```python
df = load_data()
print(f"Dataset Shape: {df.shape}")
df.head()
```

### 2. Missing Value Analysis & Summary Statistics
Inspecting data types and identifying missing entries across features.

```python
df.info()
print('\nMissing Values Count per Column:')
print(df.isnull().sum())
df.describe()
```

### 3. Target Distribution: `median_house_value`
Analyzing the target distribution reveals a capped upper limit at **$500,001**, which appears as a sharp spike on the far right.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(df['median_house_value'], kde=True, ax=axes[0], color='#2c7fb8')
axes[0].set_title('Distribution of Median House Value ($)', fontweight='bold')
sns.boxplot(x=df['median_house_value'], ax=axes[1], color='#7fcdbb')
axes[1].set_title('Boxplot of Median House Value ($)', fontweight='bold')
plt.tight_layout()
plt.show()
```

### 4. Categorical Feature Analysis: `ocean_proximity`
Examining the distribution of categorical ocean proximity values across block groups.

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='ocean_proximity', palette='Blues_r')
plt.title('House Count by Ocean Proximity', fontweight='bold')
plt.show()
```

### 5. Geographical Distribution & Price Clustering
Plotting latitude and longitude mapped to median house values clearly shows high price clusters along coastal metro areas (San Francisco Bay Area, Los Angeles, San Diego).

```python
plt.figure(figsize=(10, 7))
scatter = plt.scatter(
    df['longitude'], df['latitude'],
    c=df['median_house_value'], cmap='jet', alpha=0.4,
    s=df['population'] / 100, label='Population'
)
plt.colorbar(scatter, label='Median House Value ($)')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.title('Geographical Distribution of Housing Prices in California', fontweight='bold')
plt.legend()
plt.show()
```

### 6. Numeric Feature Correlation Analysis
`median_income` exhibits the strongest linear correlation (+0.69) with `median_house_value`.

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, fmt='.2f', cmap='coolwarm', square=True)
plt.title('Numeric Feature Correlation Heatmap', fontweight='bold')
plt.show()
```
